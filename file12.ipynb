```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
data=pd.read_csv('uber-raw-data-apr14.csv')
print(data.head())
print(data.info())
print(data.shape)
```

```python
data['Date/Time'] = pd.to_datetime(data['Date/Time'], format='%m/%d/%Y %H:%M:%S')
print(data.head())
```

```python
data["Hour"] = data['Date/Time'].dt.hour
print(data.head())
```

```python
data['Day of Week'] = data['Date/Time'].dt.dayofweek
print(data.head())
```

```python
data['Day'] = data['Date/Time'].dt.day_name()
print(data.head())
```

```python
data['Month'] = data['Date/Time'].dt.month
print(data.head())
```

```python
data['Year'] = data['Date/Time'].dt.year
print(data.head())
```

```python
data['Decade'] = data['Year'].apply(lambda x: 2010 if x < 2010 else 2020)
print(data.head())
```

```python
data['Century'] = data['Year'].apply(lambda x: 20 if x < 2000 else 21)
print(data.head())
```

```python
hourly_counts = data['Hour'].value_counts().sort_index()
print("The hour with the most rides is: Hour", hourly_counts.idxmax())
num_pickups = hourly_counts.max()
print("The number of pickups during that hour is:", num_pickups,"pickups")
```

```python
plt.plot(hourly_counts.index, hourly_counts.values)
plt.title('Number of Pickups by Hour')
plt.xlabel('Hour of the Day')
plt.ylabel('Number of Pickups')
plt.show()
```

```python
weekday_index = {0: 'Monday', 1: 'Tuesday', 2: 'Wednesday', 3: 'Thursday', 4: 'Friday', 5: 'Saturday', 6: 'Sunday'}
daily_demand = data['Day of Week'].value_counts().sort_index()
plt.bar(daily_demand.index.map(weekday_index), daily_demand.values)
plt.title('Number of Pickups by Day of the Week')
plt.xlabel('Day of the Week')
plt.ylabel('Number of Pickups')
plt.show()
```
